# Text Embedding Exploration
This notebook demonstrates several text embedding techniques on a small collection of documents. Each step is explained and visualized for clarity.

## 1. Create a Collection of Documents
Let's start by defining a small set of text documents.

```python
documents = [
    "The quick brown fox jumps over the lazy dog.",
    "Artificial intelligence is transforming the world.",
    "Natural language processing enables computers to understand text.",
    "Machine learning is a subset of AI.",
    "Dogs are loyal and friendly animals.",
    "Text embeddings are useful for many NLP tasks."
]
for i, doc in enumerate(documents):
    print(f"Doc {i+1}: {doc}")
```

## 2. Preprocessing: Lowercase, Cleaning, Remove Stopwords
We will lowercase the text, remove punctuation, and filter out stopwords.

```python
import re
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    tokens = text.split()
    tokens = [t for t in tokens if t not in stop_words]
    return tokens

preprocessed_docs = [preprocess(doc) for doc in documents]
for i, tokens in enumerate(preprocessed_docs):
    print(f"Doc {i+1} tokens:", tokens)
```

## 3. Bag of Words Construction
Let's manually build a Bag of Words representation for our documents.

```python
from collections import Counter

vocab = sorted(set([word for doc in preprocessed_docs for word in doc]))
print("Vocabulary:", vocab)

bow_matrix = []
for doc in preprocessed_docs:
    bow = [doc.count(word) for word in vocab]
    bow_matrix.append(bow)

import pandas as pd
bow_df = pd.DataFrame(bow_matrix, columns=vocab)
bow_df.index = [f"Doc {i+1}" for i in range(len(documents))]
bow_df
```

## 4. TF-IDF Construction
Now, let's manually compute the TF-IDF values for each word in each document.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Join tokens back to string for TF-IDF
docs_joined = [' '.join(doc) for doc in preprocessed_docs]
tfidf = TfidfVectorizer(vocabulary=vocab)
tfidf_matrix = tfidf.fit_transform(docs_joined).toarray()
tfidf_df = pd.DataFrame(tfidf_matrix, columns=vocab)
tfidf_df.index = [f"Doc {i+1}" for i in range(len(documents))]
tfidf_df
```

## 5. Extracted Vocabulary
Here is the final vocabulary extracted from our documents:

```python
print("Extracted Vocabulary:", vocab)
```

## 6. Word2Vec Embeddings & 3D Visualization
Let's train a Word2Vec model and visualize the word vectors in 3D.

```python
from gensim.models import Word2Vec
import numpy as np
import plotly.graph_objs as go

w2v_model = Word2Vec(sentences=preprocessed_docs, vector_size=10, window=2, min_count=1, workers=1, seed=42)
word_vectors = np.array([w2v_model.wv[word] for word in vocab])

# Reduce to 3D using PCA
from sklearn.decomposition import PCA
pca = PCA(n_components=3)
word_vecs_3d = pca.fit_transform(word_vectors)

fig = go.Figure()
fig.add_trace(go.Scatter3d(
    x=word_vecs_3d[:,0], y=word_vecs_3d[:,1], z=word_vecs_3d[:,2],
    mode='markers+text', text=vocab, marker=dict(size=6)
))
fig.update_layout(title='Word2Vec Embeddings (3D)', scene=dict(xaxis_title='X', yaxis_title='Y', zaxis_title='Z'))
fig.show()
```

## 7. MiniLM (Sentence Transformers) Embeddings & 3D Visualization
Now, let's use MiniLM to embed entire sentences and visualize them in 3D.

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('all-MiniLM-L6-v2')
embeddings = model.encode(documents)

# Reduce to 3D using PCA
embeddings_3d = PCA(n_components=3).fit_transform(embeddings)

fig2 = go.Figure()
fig2.add_trace(go.Scatter3d(
    x=embeddings_3d[:,0], y=embeddings_3d[:,1], z=embeddings_3d[:,2],
    mode='markers+text', text=[f"Doc {i+1}" for i in range(len(documents))], marker=dict(size=6)
))
fig2.update_layout(title='MiniLM Sentence Embeddings (3D)', scene=dict(xaxis_title='X', yaxis_title='Y', zaxis_title='Z'))
fig2.show()
```
