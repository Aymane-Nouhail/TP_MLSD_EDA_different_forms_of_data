```python
# Importer les librairies nécessaires
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import Audio
import os

# Télécharger le dataset depuis Kaggle (à exécuter une seule fois)
import kagglehub
path = kagglehub.dataset_download('ouaraskhelilrafik/tp-02-audio')
path
```

## Chargement d'un fichier audio du dataset Kaggle
Nous allons charger un fichier audio extrait du dataset Kaggle téléchargé précédemment.

```python
# Exemple : charger un fichier audio du dataset téléchargé
audio_path = os.path.join(path, 'Data', 'Data','101 - Dog','1-30226-A.ogg')  # Remplacez 'example.wav' par un fichier réel du dataset
y, sr = librosa.load(audio_path, sr=None)
```

## Taille et durée du signal audio
Affichons la taille (nombre d'échantillons) et la durée (en secondes) du signal chargé.

```python
# Afficher la taille et la durée du signal
print(f'Taille du signal : {len(y)} échantillons')
print(f'Durée : {len(y)/sr:.2f} secondes')
```

## Visualisation du waveform
Traçons le waveform (forme d'onde) du signal audio pour observer ses variations dans le temps.

```python
# Tracer le waveform
plt.figure(figsize=(12, 4))
librosa.display.waveshow(y, sr=sr)
plt.title('Waveform')
plt.xlabel('Temps (s)')
plt.ylabel('Amplitude')
plt.show()
```

## Calcul et affichage du spectrogramme
Calculons et visualisons le spectrogramme du signal audio pour observer la répartition des fréquences dans le temps.

```python
# Calculer et afficher le spectrogramme
D = librosa.amplitude_to_db(np.abs(librosa.stft(y)), ref=np.max)
plt.figure(figsize=(12, 4))
librosa.display.specshow(D, sr=sr, x_axis='time', y_axis='log')
plt.title('Spectrogramme (dB)')
plt.colorbar(format='%+2.0f dB')
plt.tight_layout()
plt.show()
```
